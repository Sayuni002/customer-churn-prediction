## Day 3 - Data Cleaning

In this section, we clean the dataset by handling missing values,
correcting data types, checking duplicates, and removing unnecessary columns.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/customer_churn.csv")
```

```python
df["TotalCharges"].dtype
```

```python
(df["TotalCharges"] == " ").sum()
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)
```

```python
df["TotalCharges"].isnull().sum()
```

```python
df[df["TotalCharges"].isnull()]
```

```python
df = df.dropna(subset=["TotalCharges"])
```

```python
df["TotalCharges"].isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df["customerID"].nunique()
```

```python
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    print(f"{col}: {df[col].unique()}")
    print()
```

```python
df.describe()
```

```python
df[["tenure", "MonthlyCharges", "TotalCharges"]].describe()
```

```python
print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())
print("\nData types:")
print(df.dtypes)
```

```python
df.to_csv("../data/customer_churn_cleaned.csv", index=False)
```

## Data Cleaning Summary

- Inspected missing and blank values.
- Converted `TotalCharges` to numeric format.
- Handled missing values in `TotalCharges`.
- Checked the dataset for duplicate records.
- Reviewed categorical values for inconsistencies.
- Preserved `customerID` for EDA but will exclude it from model features.
- Saved the cleaned dataset separately from the raw dataset.

##  Exploratory Data Analysis

In this section, we visualize customer churn patterns and explore
relationships between customer characteristics and churn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/customer_churn_cleaned.csv")
```

```python
sns.countplot(data=df, x="Churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

```python
df["Churn"].value_counts(normalize=True) * 100
```

### Observation: Customer Churn Distribution

- Approximately **73.42%** of customers did not churn, while **26.58%** of customers churned.
- The dataset contains noticeably more non-churn customers than churn customers.
- This indicates a **moderate class imbalance** in the target variable (`Churn`).
- Because of this imbalance, model performance should not be evaluated using accuracy alone. Metrics such as **Precision, Recall, F1-score, and ROC-AUC** will also be considered during model evaluation.

```python
sns.countplot(data=df, x="Contract", hue="Churn")

plt.title("Customer Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=15)

plt.show()
```

```python
pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100
```

```python
sns.countplot(data=df, x="gender", hue="Churn")

plt.title("Customer Churn by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")

plt.show()
```

```python
sns.countplot(data=df, x="SeniorCitizen", hue="Churn")

plt.title("Customer Churn by Senior Citizen Status")
plt.xlabel("Senior Citizen")
plt.ylabel("Number of Customers")

plt.show()
```

```python
sns.countplot(data=df, x="Partner", hue="Churn")

plt.title("Customer Churn by Partner Status")
plt.xlabel("Partner")
plt.ylabel("Number of Customers")

plt.show()
```

```python
sns.countplot(data=df, x="Dependents", hue="Churn")

plt.title("Customer Churn by Dependents")
plt.xlabel("Dependents")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="PaymentMethod",
    hue="Churn"
)

plt.title("Customer Churn by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)

plt.show()
```

```python
pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100
```

##  Key Observations

1. The dataset contains more non-churn customers than churn customers.
2. Contract type shows differences in churn patterns.
3. Churn patterns vary across payment methods.
4. Demographic variables such as gender, senior citizen status,
   partner status, and dependents were also examined.

These observations describe associations in the dataset and do not by
themselves establish that these factors cause churn.
