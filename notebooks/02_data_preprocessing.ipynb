# Customer Churn Prediction - Data Preprocessing

##  Preparing Data for Machine Learning

This notebook prepares the cleaned customer churn dataset for machine
learning by separating features and target variables, encoding categorical
features, and creating training and testing datasets.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("../data/customer_churn_cleaned.csv")

df.head()
```

```python
df.shape
```

```python
df = df.drop("customerID", axis=1)
```

```python
df.columns
```

```python
df["Churn"] = df["Churn"].map({
    "No": 0,
    "Yes": 1
})
```

```python
df["Churn"].value_counts()
```

```python
X = df.drop("Churn", axis=1)
y = df["Churn"]
```

```python
X.shape
```

```python
y.shape
```

```python
categorical_cols = X.select_dtypes(
    include="object"
).columns.tolist()

numerical_cols = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()
```

```python
print("Categorical columns:")
print(categorical_cols)

print("\nNumerical columns:")
print(numerical_cols)
```

```python
X = pd.get_dummies(
    X,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)
```

```python
X.head()
```

```python
X.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
scaler = StandardScaler()

X_train[numerical_cols] = scaler.fit_transform(
    X_train[numerical_cols]
)

X_test[numerical_cols] = scaler.transform(
    X_test[numerical_cols]
)
```

```python
print("Missing values in X_train:",
      X_train.isnull().sum().sum())

print("Missing values in X_test:",
      X_test.isnull().sum().sum())

print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

## Preprocessing Summary

- Removed the customer identifier from the model features.
- Converted the churn target into binary values (0 and 1).
- Separated features and target variables.
- Identified numerical and categorical features.
- Applied one-hot encoding to categorical variables.
- Split the dataset into 80% training and 20% testing sets.
- Used stratification to preserve the churn class distribution.
- Standardized numerical features using training-set statistics to avoid data leakage.

The dataset is now ready for machine learning model development.
